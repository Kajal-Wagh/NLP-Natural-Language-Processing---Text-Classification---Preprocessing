### N-gram

```python
An N-gram in Natural Language Processing (NLP) is a contiguous sequence of $N$ items (words, characters, or symbols) 
extracted from a given text or speech corpus.N-grams are fundamental statistical tools used to analyze language patterns, 
predict words, and process textual data.
Common N-gram TypesThe value of $N$ determines the size of the window used to group text:
    1) Unigram ($N=1$): Single words processed independently.
    Example: "Data", "science", "is", "fun"
    2)Bigram ($N=2$): Pairs of consecutive words.
    Example: "Data science", "science is", "is fun"
    3)Trigram ($N=3$): Triplets of consecutive words.
    Example: "Data science is", "science is fun"
    4-gram / $N$-gram ($N \ge 4$): Sequences of $N$ words.
    Example (4-gram): "Data science is fun"
```

```python
ord_data={'sadness':0, 'anger':1, 'love':2, 'surprise':3, 'fear':4, 'joy':5}
```

```python
df['emotions']=df['emotions'].map(ord_data)
```

```python
df.head(10)
```

```python
df['text'] = df['text'].apply(lambda x:x.lower())
```

```python
df
```

```python
import string

def remove_punc(txt):
    if isinstance(txt, str):
        return txt.translate(str.maketrans("", "", string.punctuation))
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
df.head(5)
```

```python
import re

def remove_number(txt):
  return re.sub(r"\d+", "", str(txt))

df["text"] = df["text"].apply(remove_number)
```

```python
df['text']=df['text'].apply(remove_number)
```

```python
df.head(5)
```

```python
def remove_emoji(txt):
  return "".join([i for i in str(txt) if i.isascii()])
```

```python
df['text']=df['text'].apply(remove_emoji)
```

```python
df.head(5)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
stop_word=set(stopwords.words('english'))
```

```python
stop_word
```

```python
df.loc[1]['text']
```

```python
def remove(txt):
    word = word_tokenize(txt)
    cleaned = []
    for i in word:
        if i not in stop_word:
            cleaned.append(i)
    return ' '.join(cleaned)
```

```python
df['text']=df['text'].apply(remove)
```

```python
reverse_map={
    0:'sadness', 
    1:'anger', 
    2:'love', 
    3:'surprise', 
    4:'fear',
    5:'joy'
}
```

```python
def predict_emotions(text):
    text=text.lower()

    #remove_punctuation
    text= remove_punc(text)
    text= remove_number(text)

    #remove_emoji
    text=remove_emoji(text)
    
    #remove stopword
    text=remove(text)

    #convert text into vector
    text_vector=cv.transform([text])

    #predict
    pred=int(model.predict(text_vector)[0])

    #convert number bact to emotions
    return reverse_map[pred]
    
```

```python
new_text= "i am feeling 5 very happy 😊 today!"
```

```python
result=predict_emotions(new_text)
```

```python
result
```
